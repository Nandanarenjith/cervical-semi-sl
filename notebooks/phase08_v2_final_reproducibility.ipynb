```python
from pathlib import Path
import json

PROJECT_ROOT = Path.cwd().parent

CONFIG_PATH = PROJECT_ROOT / "configs" / "phase8_v2_reproducibility.json"
SPLIT_ROOT = PROJECT_ROOT / "data" / "splits_v2"

assert CONFIG_PATH.exists(), f"Missing config: {CONFIG_PATH}"
assert SPLIT_ROOT.exists(), f"Missing split directory: {SPLIT_ROOT}"

with open(CONFIG_PATH, "r") as f:
    config = json.load(f)

print("Config loaded successfully")
print("Protocol:", config["protocol_version"])
print("Split version:", config["split_version"])
print("Seed:", config["seed"])
```

```python
assert config["protocol_version"] == "FINAL_V2"
assert config["split_version"] == "V2"
assert config["seed"] == 42

assert config["split"]["train"] == 0.70
assert config["split"]["validation"] == 0.10
assert config["split"]["test"] == 0.20

assert sum(config["split"].values()) == 1.0

assert config["datasets"]["cric"]["group"] == "image"
assert config["datasets"]["riva"]["group"] == "slide"

assert config["image"]["size"] == 128
assert config["image"]["channels"] == 3

assert config["labels"]["0"] == "normal_non_pathological"
assert config["labels"]["1"] == "abnormal_pathological"

assert config["label_fractions"] == [0.05, 0.10]

print("ALL FINAL V2 PROTOCOL CHECKS PASSED")
```

```python
CRIC_SPLIT = SPLIT_ROOT / "cric_splits_v2.csv"
RIVA_SPLIT = SPLIT_ROOT / "riva_splits_v2.csv"

assert CRIC_SPLIT.exists()
assert RIVA_SPLIT.exists()

print("CRIC split:", CRIC_SPLIT)
print("RIVA split:", RIVA_SPLIT)
print()
print("PHASE 08 V2 — COMPLETE")
print("Final protocol is locked and reproducible.")
```
