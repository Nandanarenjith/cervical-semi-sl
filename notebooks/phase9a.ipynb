```python
import torch
import torchvision
import time

from torchvision.models import (
    resnet18,
    ResNet18_Weights,
    densenet121,
    DenseNet121_Weights,
    efficientnet_b0,
    EfficientNet_B0_Weights,
    convnext_tiny,
    ConvNeXt_Tiny_Weights,
    swin_t,
    Swin_T_Weights,
    vit_b_16,
    ViT_B_16_Weights,
)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("DEVICE:", DEVICE)
print("GPU:", torch.cuda.get_device_name(0))
print()
```

```python
models_to_test = {
    "ResNet-18": resnet18(weights=None),
    "DenseNet-121": densenet121(weights=None),
    "EfficientNet-B0": efficientnet_b0(weights=None),
    "ConvNeXt-Tiny": convnext_tiny(weights=None),
    "Swin-Tiny": swin_t(weights=None),
    "ViT-B/16": vit_b_16(weights=None),
}

dummy_input = torch.randn(2, 3, 224, 224)

results = []

for name, model in models_to_test.items():

    # Replace classification head BEFORE moving model to GPU
    if name == "ResNet-18":
        model.fc = torch.nn.Linear(
            model.fc.in_features, 2
        )

    elif name == "DenseNet-121":
        model.classifier = torch.nn.Linear(
            model.classifier.in_features, 2
        )

    elif name == "EfficientNet-B0":
        model.classifier[1] = torch.nn.Linear(
            model.classifier[1].in_features, 2
        )

    elif name == "ConvNeXt-Tiny":
        model.classifier[2] = torch.nn.Linear(
            model.classifier[2].in_features, 2
        )

    elif name == "Swin-Tiny":
        model.head = torch.nn.Linear(
            model.head.in_features, 2
        )

    elif name == "ViT-B/16":
        model.heads.head = torch.nn.Linear(
            model.heads.head.in_features, 2
        )

    # NOW move the complete model to GPU
    model = model.to(DEVICE)
    model.eval()

    # Move input to GPU
    input_gpu = dummy_input.to(DEVICE)

    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()

    start = time.time()

    with torch.no_grad():
        output = model(input_gpu)
        torch.cuda.synchronize()

    elapsed = time.time() - start

    peak_memory = torch.cuda.max_memory_allocated() / (1024 ** 3)

    param_count = sum(
        p.numel() for p in model.parameters()
    ) / 1e6

    results.append({
        "backbone": name,
        "output_shape": tuple(output.shape),
        "parameters_M": round(param_count, 2),
        "forward_time_sec": round(elapsed, 3),
        "peak_GPU_memory_GB": round(peak_memory, 3),
    })

    print(
        f"{name:18s} | "
        f"output={tuple(output.shape)} | "
        f"params={param_count:.2f}M | "
        f"time={elapsed:.3f}s | "
        f"VRAM={peak_memory:.3f} GB"
    )

    del model
    del input_gpu
    torch.cuda.empty_cache()
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image

class CRICDataset(Dataset):
    def __init__(self, dataframe, crop_root, transform=None):
        self.df = dataframe.reset_index(drop=True)
        self.crop_root = crop_root
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        crop_filename = Path(row["crop_path"]).name
        image_path = self.crop_root / crop_filename

        image = Image.open(image_path).convert("RGB")
        label = int(row["binary_label"])

        if self.transform:
            image = self.transform(image)

        return image, label


# Use the fixed Phase 8 split
train_df = cric_splits[cric_splits["split"] == "train"].copy()
val_df = cric_splits[cric_splits["split"] == "val"].copy()

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

train_dataset = CRICDataset(
    train_df,
    CRIC_CROPS,
    transform=transform
)

val_dataset = CRICDataset(
    val_df,
    CRIC_CROPS,
    transform=transform
)

BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=True
)

images, labels = next(iter(train_loader))

print("TRAIN DATASET:", len(train_dataset))
print("VAL DATASET:", len(val_dataset))
print("IMAGE BATCH SHAPE:", images.shape)
print("LABEL BATCH SHAPE:", labels.shape)
print("LABELS:", labels[:10].tolist())
print("IMAGE DEVICE BEFORE TRAINING:", images.device)
```

```python
import torch
import torch.nn as nn

from torchvision.models import (
    resnet18,
    ResNet18_Weights,
    densenet121,
    DenseNet121_Weights,
    efficientnet_b0,
    EfficientNet_B0_Weights,
    convnext_tiny,
    ConvNeXt_Tiny_Weights,
    swin_t,
    Swin_T_Weights,
    vit_b_16,
    ViT_B_16_Weights,
)

NUM_CLASSES = 2

def build_model(name):
    if name == "ResNet-18":
        model = resnet18(weights=ResNet18_Weights.DEFAULT)
        model.fc = nn.Linear(model.fc.in_features, NUM_CLASSES)

    elif name == "DenseNet-121":
        model = densenet121(weights=DenseNet121_Weights.DEFAULT)
        model.classifier = nn.Linear(
            model.classifier.in_features,
            NUM_CLASSES
        )

    elif name == "EfficientNet-B0":
        model = efficientnet_b0(weights=EfficientNet_B0_Weights.DEFAULT)
        model.classifier[1] = nn.Linear(
            model.classifier[1].in_features,
            NUM_CLASSES
        )

    elif name == "ConvNeXt-Tiny":
        model = convnext_tiny(weights=ConvNeXt_Tiny_Weights.DEFAULT)
        model.classifier[2] = nn.Linear(
            model.classifier[2].in_features,
            NUM_CLASSES
        )

    elif name == "Swin-Tiny":
        model = swin_t(weights=Swin_T_Weights.DEFAULT)
        model.head = nn.Linear(
            model.head.in_features,
            NUM_CLASSES
        )

    elif name == "ViT-B/16":
        model = vit_b_16(weights=ViT_B_16_Weights.DEFAULT)
        model.heads.head = nn.Linear(
            model.heads.head.in_features,
            NUM_CLASSES
        )

    else:
        raise ValueError(f"Unknown model: {name}")

    return model.to(DEVICE)


candidate_names = [
    "ResNet-18",
    "DenseNet-121",
    "EfficientNet-B0",
    "ConvNeXt-Tiny",
    "Swin-Tiny",
    "ViT-B/16"
]

print("Candidate models:")
for name in candidate_names:
    print("-", name)
```

```python
from torchvision import transforms
from torch.utils.data import DataLoader

# Base normalization
normalize = transforms.Normalize(
    mean=[0.485, 0.456, 0.406],
    std=[0.229, 0.224, 0.225]
)

# 128x128 input for CNNs and Swin-Tiny
transform_128 = transforms.Compose([
    transforms.ToTensor(),
    normalize
])

# 224x224 input required by ViT-B/16
transform_224 = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    normalize
])

# Recreate datasets
train_dataset_128 = CRICDataset(
    train_df,
    CRIC_CROPS,
    transform=transform_128
)

val_dataset_128 = CRICDataset(
    val_df,
    CRIC_CROPS,
    transform=transform_128
)

train_dataset_224 = CRICDataset(
    train_df,
    CRIC_CROPS,
    transform=transform_224
)

val_dataset_224 = CRICDataset(
    val_df,
    CRIC_CROPS,
    transform=transform_224
)

# DataLoaders
train_loader_128 = DataLoader(
    train_dataset_128,
    batch_size=32,
    shuffle=True,
    num_workers=0,
    pin_memory=True
)

val_loader_128 = DataLoader(
    val_dataset_128,
    batch_size=32,
    shuffle=False,
    num_workers=0,
    pin_memory=True
)

train_loader_224 = DataLoader(
    train_dataset_224,
    batch_size=32,
    shuffle=True,
    num_workers=0,
    pin_memory=True
)

val_loader_224 = DataLoader(
    val_dataset_224,
    batch_size=32,
    shuffle=False,
    num_workers=0,
    pin_memory=True
)

# Verify both
images_128, labels_128 = next(iter(train_loader_128))
images_224, labels_224 = next(iter(train_loader_224))

print("128x128 batch:", images_128.shape)
print("224x224 batch:", images_224.shape)
print("Labels:", labels_128.shape)
```
